The data in this notebook is sourced from:
     Kielland, K., F.S. Chapin, R.W. Ruess, and Bonanza Creek LTER. 2017. Snowshoe hare physical data in Bonanza Creek Experimental Forest: 1999-Present ver 22. Environmental Data Initiative. https://doi.org/10.6073/pasta/03dce4856d79b91557d8e6ce2cbcdc14 (Accessed 2026-10-16).

https://portal.edirepository.org/nis/mapbrowse?packageid=knb-lter-bnz.55.22 

The data was assembled by Kielland, Chapin, and Ruess, as well as the Bonanza Creek LTER. The purpose of the collection of the data was to study the snowshoe hare population over a period of time. 

A singular row represents one trapping of one snowshoe hare.

The observations were collected between 1999 and 2012, between Tok and Clear, Alaska. 

There is no sensitive data.

The study protocols included trapping the snowhares, as well as tagging the right and left ears. 

Weight is measured in grams. 

Missing data is encoded as a blank cell.

The associated publication is Kielland, K., F.S. Chapin, R.W. Ruess, and Bonanza Creek LTER. 2017. Snowshoe hare physical data in Bonanza Creek Experimental Forest: 1999-Present ver 22. Environmental Data Initiative. https://doi.org/10.6073/pasta/03dce4856d79b91557d8e6ce2cbcdc14 (Accessed 2026-10-16).
